```python
# ═══════════════════════════════════════════════════════════════
# CONVERTIR ARCHIVOS A TABLAS DE UNITY CATALOG (CORREGIDO)
# Ejecutar este código en Databricks para crear tablas
# ═══════════════════════════════════════════════════════════════

from pyspark.sql import SparkSession
from pyspark.sql import functions as F

spark = SparkSession.builder.getOrCreate()

# Configuración
DASHBOARD_PATH = "/Volumes/olist/olist_gold/dashboard/"
CATALOG = "olist"
SCHEMA = "olist_gold"

print("="*80)
print("🔄 CONVIRTIENDO ARCHIVOS A TABLAS DE UNITY CATALOG")
print("="*80)
print()

# ═══════════════════════════════════════════════════════════════
# TABLA 1: kmeans_metrics
# ═══════════════════════════════════════════════════════════════

print("1️⃣ Creando tabla: kmeans_metrics")

df_kmeans = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load(f"{DASHBOARD_PATH}kmeans_metrics.csv")

df_kmeans.write.format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable(f"{CATALOG}.{SCHEMA}.kmeans_metrics")

print(f"   ✅ Tabla creada: {CATALOG}.{SCHEMA}.kmeans_metrics")
print(f"      Registros: {df_kmeans.count()}")
print()

# ═══════════════════════════════════════════════════════════════
# TABLA 2: customers_segmented
# ═══════════════════════════════════════════════════════════════

print("2️⃣ Creando tabla: customers_segmented")

df_customers = spark.read.format("delta") \
    .load(f"{DASHBOARD_PATH}customers_segmented/")

df_customers.write.format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable(f"{CATALOG}.{SCHEMA}.customers_segmented")

print(f"   ✅ Tabla creada: {CATALOG}.{SCHEMA}.customers_segmented")
print(f"      Registros: {df_customers.count()}")
print()

# ═══════════════════════════════════════════════════════════════
# TABLA 3: customer_features
# ═══════════════════════════════════════════════════════════════

print("3️⃣ Creando tabla: customer_features")

df_features = spark.read.format("delta") \
    .load(f"{DASHBOARD_PATH}customer_features/")

df_features.write.format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable(f"{CATALOG}.{SCHEMA}.customer_features_dashboard")

print(f"   ✅ Tabla creada: {CATALOG}.{SCHEMA}.customer_features_dashboard")
print(f"      Registros: {df_features.count()}")
print()

# ═══════════════════════════════════════════════════════════════
# TABLA 4: correlation_matrix
# ═══════════════════════════════════════════════════════════════

print("4️⃣ Creando tabla: correlation_matrix")

df_corr = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load(f"{DASHBOARD_PATH}correlation_matrix.csv")

df_corr.write.format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable(f"{CATALOG}.{SCHEMA}.correlation_matrix")

print(f"   ✅ Tabla creada: {CATALOG}.{SCHEMA}.correlation_matrix")
print(f"      Registros: {df_corr.count()}")
print()

# ═══════════════════════════════════════════════════════════════
# TABLA 5: features_eliminadas
# ═══════════════════════════════════════════════════════════════

print("5️⃣ Creando tabla: features_eliminadas")

try:
    df_feat_elim = spark.read.format("csv") \
        .option("header", "true") \
        .load(f"{DASHBOARD_PATH}features_eliminadas.csv")
    
    df_feat_elim.write.format("delta") \
        .mode("overwrite") \
        .option("overwriteSchema", "true") \
        .saveAsTable(f"{CATALOG}.{SCHEMA}.features_eliminadas")
    
    print(f"   ✅ Tabla creada: {CATALOG}.{SCHEMA}.features_eliminadas")
    print(f"      Registros: {df_feat_elim.count()}")
except:
    print(f"   ⚠️ No hay features eliminadas")
print()

# ═══════════════════════════════════════════════════════════════
# TABLA 6: pca_params
# ═══════════════════════════════════════════════════════════════

print("6️⃣ Creando tabla: pca_params")

df_pca = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load(f"{DASHBOARD_PATH}pca_params.csv")

df_pca.write.format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable(f"{CATALOG}.{SCHEMA}.pca_params")

print(f"   ✅ Tabla creada: {CATALOG}.{SCHEMA}.pca_params")
print(f"      Registros: {df_pca.count()}")
print()

# ═══════════════════════════════════════════════════════════════
# TABLA 7: model_comparison
# ═══════════════════════════════════════════════════════════════

print("7️⃣ Creando tabla: model_comparison")

df_models = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load(f"{DASHBOARD_PATH}model_comparison.csv")

df_models.write.format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable(f"{CATALOG}.{SCHEMA}.model_comparison")

print(f"   ✅ Tabla creada: {CATALOG}.{SCHEMA}.model_comparison")
print(f"      Registros: {df_models.count()}")
print()

# ═══════════════════════════════════════════════════════════════
# TABLA 8: confusion_matrix (CORREGIDA)
# ═══════════════════════════════════════════════════════════════

print("8️⃣ Creando tabla: confusion_matrix")

df_conf = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load(f"{DASHBOARD_PATH}confusion_matrix.csv")

# CORRECCIÓN: Renombrar columnas eliminando espacios y caracteres especiales
for col_name in df_conf.columns:
    new_col_name = col_name.replace(" ", "_").replace("-", "_")
    df_conf = df_conf.withColumnRenamed(col_name, new_col_name)

df_conf.write.format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable(f"{CATALOG}.{SCHEMA}.confusion_matrix_validation")

print(f"   ✅ Tabla creada: {CATALOG}.{SCHEMA}.confusion_matrix_validation")
print(f"      Registros: {df_conf.count()}")
print()

# ═══════════════════════════════════════════════════════════════
# TABLA 9: test_metrics
# ═══════════════════════════════════════════════════════════════

print("9️⃣ Creando tabla: test_metrics")

df_test_metrics = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load(f"{DASHBOARD_PATH}test_metrics.csv")

df_test_metrics.write.format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable(f"{CATALOG}.{SCHEMA}.test_metrics")

print(f"   ✅ Tabla creada: {CATALOG}.{SCHEMA}.test_metrics")
print(f"      Registros: {df_test_metrics.count()}")
print()

# ═══════════════════════════════════════════════════════════════
# TABLA 10: test_confusion_matrix (CORREGIDA)
# ═══════════════════════════════════════════════════════════════

print("🔟 Creando tabla: test_confusion_matrix")

df_test_conf = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load(f"{DASHBOARD_PATH}test_confusion_matrix.csv")

# CORRECCIÓN: Renombrar columnas eliminando espacios
for col_name in df_test_conf.columns:
    new_col_name = col_name.replace(" ", "_").replace("-", "_")
    df_test_conf = df_test_conf.withColumnRenamed(col_name, new_col_name)

df_test_conf.write.format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable(f"{CATALOG}.{SCHEMA}.test_confusion_matrix")

print(f"   ✅ Tabla creada: {CATALOG}.{SCHEMA}.test_confusion_matrix")
print(f"      Registros: {df_test_conf.count()}")
print()

# ═══════════════════════════════════════════════════════════════
# TABLA 11: inference_dataset_info
# ═══════════════════════════════════════════════════════════════

print("1️⃣1️⃣ Creando tabla: inference_dataset_info")

df_inf_info = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load(f"{DASHBOARD_PATH}inference_dataset_info.csv")

df_inf_info.write.format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable(f"{CATALOG}.{SCHEMA}.inference_dataset_info")

print(f"   ✅ Tabla creada: {CATALOG}.{SCHEMA}.inference_dataset_info")
print(f"      Registros: {df_inf_info.count()}")
print()

# ═══════════════════════════════════════════════════════════════
# TABLA 12: inference_dataset_sample
# ═══════════════════════════════════════════════════════════════

print("1️⃣2️⃣ Creando tabla: inference_dataset_sample")

df_inf_sample = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load(f"{DASHBOARD_PATH}inference_dataset_sample.csv")

df_inf_sample.write.format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable(f"{CATALOG}.{SCHEMA}.inference_dataset_sample")

print(f"   ✅ Tabla creada: {CATALOG}.{SCHEMA}.inference_dataset_sample")
print(f"      Registros: {df_inf_sample.count()}")
print()

# ═══════════════════════════════════════════════════════════════
# TABLA 13: inference_results
# ═══════════════════════════════════════════════════════════════

print("1️⃣3️⃣ Creando tabla: inference_results")

df_inf_results = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load(f"{DASHBOARD_PATH}inference_results_sample.csv")

df_inf_results.write.format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable(f"{CATALOG}.{SCHEMA}.inference_results")

print(f"   ✅ Tabla creada: {CATALOG}.{SCHEMA}.inference_results")
print(f"      Registros: {df_inf_results.count()}")
print()

# ═══════════════════════════════════════════════════════════════
# RESUMEN
# ═══════════════════════════════════════════════════════════════

print("="*80)
print("✅ TODAS LAS TABLAS CREADAS EXITOSAMENTE")
print("="*80)
print()

print("📊 TABLAS DISPONIBLES EN UNITY CATALOG:")
print()
print(f"   {CATALOG}.{SCHEMA}.kmeans_metrics")
print(f"   {CATALOG}.{SCHEMA}.customers_segmented")
print(f"   {CATALOG}.{SCHEMA}.customer_features_dashboard")
print(f"   {CATALOG}.{SCHEMA}.correlation_matrix")
print(f"   {CATALOG}.{SCHEMA}.features_eliminadas")
print(f"   {CATALOG}.{SCHEMA}.pca_params")
print(f"   {CATALOG}.{SCHEMA}.model_comparison")
print(f"   {CATALOG}.{SCHEMA}.confusion_matrix_validation")
print(f"   {CATALOG}.{SCHEMA}.test_metrics")
print(f"   {CATALOG}.{SCHEMA}.test_confusion_matrix")
print(f"   {CATALOG}.{SCHEMA}.inference_dataset_info")
print(f"   {CATALOG}.{SCHEMA}.inference_dataset_sample")
print(f"   {CATALOG}.{SCHEMA}.inference_results")
print()

print("🔍 VERIFICAR:")
print("   spark.sql('SHOW TABLES IN olist.olist_gold').show()")
print()

print("📋 EJEMPLO DE CONSULTA:")
print("   df = spark.sql('SELECT * FROM olist.olist_gold.kmeans_metrics').toPandas()")
print()

print("✅ Las columnas con espacios fueron renombradas automáticamente:")
print("   'No Premium' → 'No_Premium'")
print("   'Premium' → 'Premium'")
print()

print("="*80)
```
